# Packages

```python
import pandas as pd
import numpy as np
import gc
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.decomposition import PCA
from sklearn.linear_model import LinearRegression
from sklearn.svm import LinearSVR
from sklearn.impute import KNNImputer, SimpleImputer
from sklearn.metrics import r2_score, root_mean_squared_error
import sklearn

import sys

RANDOM_STATE = 11

sklearn.random.seed(RANDOM_STATE)
np.random.seed(RANDOM_STATE)
```

```python
def check_or_create_directory(dir_path:str="./outputs"):
    from pathlib import Path
    # creating a new directory
    path = Path(dir_path)
    if not path.exists():
        path.mkdir(parents=True, exist_ok=True)
        print("Created directory: {}".format(path))
    else:
        print("Directory already exists: {}".format(path))
```

# Loading Data 

```python
df = pd.read_excel("./FFS studies metadata.xlsx", sheet_name="Data extraction")
print(df.shape)
df
```

```python
df = df.loc[:27, ["#Datasets", "#Baselines", "#NewMethods", "WinRate"]]
gc.collect()
print(df.shape)
df
```

```python
df = df.replace("-", np.nan)
df.isna().sum()
```

# Preprocessing

```python
# # Dropping missing values 
# df = df.dropna(
#     subset=[
#         "WinRate",
#         "#Datasets",
#     ]
# ).reset_index(drop=True)
# print(df.shape)
# df
```

```python
# pd.DataFrame.fillna?
```

```python
# Mean imputer
for col in df.columns:
    df[col] = df[col].fillna(value=df[col].mean())
    # print(df[col].mean())
```

# Splitting data into train and test

```python
X = df.drop(
    [
        "WinRate",
        # "#Datasets", 
        # "#Baselines", 
        # "#NewMethods",
    ], 
    axis=1
)
y = df["WinRate"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33, shuffle=True,
    random_state=RANDOM_STATE
)

print(X_train.shape, X_test.shape)
```

# Training a linear model 

## Using train and test data

```python
lr = LinearRegression()
lr.fit(X_train, y_train)
print("Regression coef.: ", lr.coef_)
```

```python
y_train_p = lr.predict(X_train)
y_test_p = lr.predict(X_test)

r2_score(y_train, y_train_p) , r2_score(y_test, y_test_p)
```

## Using the whole data

```python
lr = LinearRegression()
lr.fit(X, y)
```

```python
y_p = lr.predict(X)

rmse = root_mean_squared_error(y, y_p) 
r2 = lr.score(X,y)

rmse, r2
```

## PCA components

```python
pca = PCA(n_components=0.99)
pca.fit(X)
pca.explained_variance_ratio_
```

```python
pca = PCA(n_components=1)
pca.fit(X)
pca.explained_variance_ratio_
```

```python
X_pca = pca.transform(X)
```

## Using the PCA components

```python
lr = LinearRegression()
lr.fit(X_pca, y)
```

```python
y_pca_p = lr.predict(X_pca)

rmse_pca = root_mean_squared_error(y, y_pca_p) 
r2_pca2 = r2_score(y, y_pca_p) 

r2_pca =lr.score(X_pca,y)

rmse_pca, r2_pca , r2_pca2
```

# Visualization

```python
imgs_saving_dir = "./outputs/imgs"
check_or_create_directory(imgs_saving_dir)
```

```python
plt.style.use('default')
plt.style.use('ggplot')

fig, ax = plt.subplots(figsize=(8, 4))

ax.plot(X_pca, y_pca_p, color='k', label='Linear Regression')
ax.scatter(X_pca, y, edgecolor='k', facecolor='grey', alpha=0.7, label='Sample data')
ax.set_ylabel('Win Rate', fontsize=14, color='k')
ax.set_xlabel('1 PCA component with {:.2%} of the Explained Variance of the Sample data'.format(pca.explained_variance_ratio_[0]), fontsize=12, color='k')
ax.legend(facecolor='white', fontsize=11)
ax.set_title('$R^2= {:.2f}, RMSE={:.2f}$'.format(r2, rmse) , fontsize=16)

fig.tight_layout()
fig.savefig(imgs_saving_dir+"/correlation between Win Rate and Sample data.png", dpi=600, bbox_inches='tight')
```

# Uni- and bi-variate models

```python
from itertools import combinations
```

```python
plt.style.use('default')
plt.style.use('ggplot')

y = df["WinRate"]

for r in range(1,3):
    cols_combination = list(combinations(df.drop("WinRate", axis=1).columns, r))
    print(len(cols_combination))
    
    fig, ax = plt.subplots(figsize=(8, 12), nrows=len(cols_combination))
    ax = ax.flatten()
    
    for i, cols in enumerate(cols_combination):
        print(cols)
        X = df.loc[:, cols]
        print(X.shape)

        # Training Linear model
        lr = LinearRegression()
        lr.fit(X, y)
        print("Regression coef.: ", lr.coef_)
        y_p = lr.predict(X)
        
        rmse = root_mean_squared_error(y, y_p) 
        r2 = lr.score(X,y)
        print(r2, rmse)

        # pca, X_pca, y_pca_p = None, None, None
        # if r>1:
        # PCA components
        pca = PCA(n_components=1)
        pca.fit(X)
        X_pca = pca.transform(X) if X.shape[1]>1 else X # To keep the original feature values for univariate
        # Training using PCA components
        lr_pca = LinearRegression()
        lr_pca.fit(X_pca, y)
        y_pca_p = lr_pca.predict(X_pca)
        rmse_pca = root_mean_squared_error(y, y_pca_p) 
        r2_pca =lr_pca.score(X_pca,y)
        print(r2_pca, rmse_pca)
        # else:
        #     X_pca = df[cols[0]]
        #     y_pca_p = y
            
        ax[i].plot(
            X_pca, y_pca_p, color='k', label='Linear Regression')
        label_sample_data = ' and '.join(cols) if r>1 else cols[0] #'Sample data'
        ax[i].scatter(X_pca, y, edgecolor='k', facecolor='grey', alpha=0.7, label=label_sample_data )
        ax[i].set_ylabel('Win Rate', fontsize=14, color='k')
        
        xlabel = None
        if r>1:
            type_of_analysis = "Bivariate Analysis" if r==2 else "Sample Data"
            xlabel = '1 PCA component with {:.2%} of the Explained Variance of the {}'.format(pca.explained_variance_ratio_[0], type_of_analysis)
        else:
            xlabel = cols[0]
            
        ax[i].set_xlabel(xlabel, fontsize=12 if r>1 else 14, color='k')
        ax[i].legend(facecolor='white', fontsize=11)
        ax[i].set_title('$R^2= {:.2f}, RMSE={:.2f}$'.format(r2, rmse) , fontsize=16)
        
    fig.tight_layout()
    if r==1:
        fig.savefig(imgs_saving_dir+"/univariate correlation between Win Rate and Sample data.png", dpi=600, bbox_inches='tight')
    elif r==2:
        fig.savefig(imgs_saving_dir+"/bivariate correlation between Win Rate and Sample data.png", dpi=600, bbox_inches='tight')
        
```
